```python
# code to solve set of linear equation Ax=b using numpy
from scipy import optimize

a = 200
c = 170


# solve set of three nonlinear equations
def equations(vars):
    b, d = vars
    eq1 = a + b + d - 300
    eq2 = (b - 90) ** 2 - (d + c) ** 2 + a**2

    return [eq1, eq2]


b, y = 1, 1
solution = optimize.fsolve(equations, (b, y))
print(f"Solution: b={solution[0]}, d={solution[1]}")

print(solution[1] + c)
```
