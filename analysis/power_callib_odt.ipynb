```python
# CDT1 POWER VOLTAGE CALIBRATION
# Data: Power (mW) vs Voltage (mV)


import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

#  CDT1 POWER VOLTAGE CALIBRATION
# Data: Power (mW) vs Voltage (V)
powercdt1 = np.array([11, 18.8, 32, 47.5, 91, 150])  # power in mW
voltagecdt1 = np.array([0.6, 0.7, 0.85, 1.0, 1.5, 2.1]) / 10  # voltage in V


# CDT2 POWER VOLTAGE CALIBRATION
# Data: Power (mW) vs Voltage (V)

powercdt2 = np.array([0.157, 0.26, 0.455, 0.663, 1.26, 2.072]) * 1000  # power in mW
voltagecdt2 = np.array([0.49, 0.7, 1.2, 1.6, 3.0, 5.0]) / 10  # voltage in V


# Fit function: Linear
def linear_fit(x, a, b):
    return a * x + b


# Fit data
paramscdt1, covariancecdt1 = curve_fit(linear_fit, voltagecdt1, powercdt1)

# Extracting the fitting parameters
slopecdt1, interceptcdt1 = paramscdt1

# Generate fitted data for plotting
fitted_power_cdt1 = linear_fit(voltagecdt1, slopecdt1, interceptcdt1)

# Plotting the data and the linear fit
plt.figure(figsize=(8, 6))
plt.plot(voltagecdt1, powercdt1, "o", label="Data")
plt.plot(
    voltagecdt1,
    fitted_power_cdt1,
    "-",
    label=f"Linear Fit: P = {slopecdt1:.6f} * V + {interceptcdt1:.6f}",
)
plt.xlabel("Voltage (V)")
plt.ylabel("Power (mW)")
plt.title("Power vs Voltage (Photodiode)(cdt1)")
plt.grid(True)
plt.legend()
plt.show()

# Output the gain (slope)
print(f"Gain (slope): {slopecdt1:.6f}")
print(f"Intercept: {interceptcdt1:.6f}")


# Fit data for CDT2
paramscdt2, covariancecdt2 = curve_fit(linear_fit, voltagecdt2, powercdt2)

# Extracting the fitting parameters
slopecdt2, interceptcdt2 = paramscdt2

# Generate fitted data for plotting
fitted_voltage_cdt2 = linear_fit(voltagecdt2, slopecdt2, interceptcdt2)

# Plotting the data and the linear fit
plt.figure(figsize=(8, 6))
plt.plot(voltagecdt2, powercdt2, "o", label="Data")
plt.plot(
    voltagecdt2,
    fitted_voltage_cdt2,
    "-",
    label=f"Linear Fit: V = {slopecdt2:.6f} * P + {interceptcdt2:.6f}",
)
plt.xlabel("Voltage (V)")
plt.ylabel("Power (mW)")
plt.title("Power vs Voltage (Photodiode)(cdt2)")
plt.legend()
plt.grid(True)
plt.show()

# Output the gain (slope)
print(f"Gain (slope): {slopecdt2:.6f}")
print(f"Intercept: {interceptcdt2:.6f}")
```
