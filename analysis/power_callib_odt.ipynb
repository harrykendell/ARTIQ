```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import AutoMinorLocator

# ----------------------------
# Data
# ----------------------------
power_screen = np.array([1.11, 0.71, 0.50, 0.30, 0.20]) * 1000  # mW
power_on_cell = np.array([0.463, 0.317, 0.240, 0.165, 0.132]) * 1000  # mW

# ----------------------------
# Linear fit
# ----------------------------
slope, intercept = np.polyfit(power_screen, power_on_cell, 1)

fit = slope * power_screen + intercept
r2 = 1 - np.sum((power_on_cell - fit) ** 2) / np.sum(
    (power_on_cell - np.mean(power_on_cell)) ** 2
)

# Smooth line
xfit = np.linspace(power_screen.min(), power_screen.max(), 300)
yfit = slope * xfit + intercept

# ----------------------------
# Figure style
# ----------------------------
plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 12,
    "axes.linewidth": 1.2,
    "axes.labelsize": 14,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "legend.fontsize": 11,
    "pdf.fonttype": 42,  # Editable text in Illustrator
    "ps.fonttype": 42,
})

fig, ax = plt.subplots(figsize=(4.2, 3.6))

# Scatter
ax.scatter(
    power_screen,
    power_on_cell,
    s=60,
    facecolor="white",
    edgecolor="black",
    linewidth=1.4,
    zorder=3,
    label="Measurement",
)

# Linear fit
ax.plot(
    xfit,
    yfit,
    color="tab:blue",
    linewidth=2.2,
    label="Linear fit",
)

# Labels
ax.set_xlabel("Power on screen (mW)")
ax.set_ylabel("Power on cell (mW)")

# Minor ticks
ax.xaxis.set_minor_locator(AutoMinorLocator())
ax.yaxis.set_minor_locator(AutoMinorLocator())

# Tick appearance
ax.tick_params(direction="in", which="both", top=False, right=False)
ax.tick_params(which="major", length=6, width=1.2)
ax.tick_params(which="minor", length=3, width=1.0)

# Remove unnecessary spines
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Fit information
eqn = (
    rf"$P_{{\rm cell}}={slope:.3f}P_{{\rm screen}}"
    rf"{intercept:+.2f}$"
    "\n"
    rf"$R^2={r2:.4f}$"
)

ax.text(
    0.04,
    0.96,
    eqn,
    transform=ax.transAxes,
    ha="left",
    va="top",
    fontsize=11,
    bbox=dict(facecolor="white", edgecolor="none", alpha=0.9),
)

ax.legend(frameon=False, loc="lower right")

plt.tight_layout()

# Publication-quality output
plt.savefig("power_calibration.pdf", bbox_inches="tight")
plt.savefig("power_calibration.png", dpi=600, bbox_inches="tight")

plt.show()
```
