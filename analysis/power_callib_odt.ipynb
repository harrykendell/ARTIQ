```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ==================================================
# 1. CALIBRATION PARAMETERS
# ==================================================

a = 2.3361  # Calibration slope (mW/V)
b = -0.1964  # Calibration intercept (mW)

# Calibration equation:
# P(mW) = 2.3361 * V(V) - 0.1964


# ==================================================
# 2. LOAD OSCILLOSCOPE CSV FILE
# ==================================================

filename = "/home/ae19663/Desktop/Deepak/02-11-2026/power stabilitiy of dimple over short term 10s_000.csv"

with open(filename, "r", encoding="utf-8-sig") as f:
    lines = f.readlines()

# Locate the row containing TIME and CH3
header_row = None

for i, line in enumerate(lines):
    if "TIME" in line and "CH3" in line:
        header_row = i
        break

if header_row is None:
    raise ValueError("TIME and CH3 columns not found in CSV.")

print("Header found at row:", header_row)


# ==================================================
# 3. READ DATA
# ==================================================

df = pd.read_csv(filename, skiprows=header_row, sep=None, engine="python")

df.columns = df.columns.str.strip()

print("\nCSV columns:")
print(df.columns.tolist())

# Extract time and voltage
time = pd.to_numeric(df["TIME"], errors="coerce")
voltage = pd.to_numeric(df["CH3"], errors="coerce")

# Remove invalid values
valid = time.notna() & voltage.notna()

time = time[valid].to_numpy()
voltage = voltage[valid].to_numpy()

# Sort data by time
sort_index = np.argsort(time)

time = time[sort_index]
voltage = voltage[sort_index]

print("\nNumber of samples:", len(time))


# ==================================================
# 4. CONVERT VOLTAGE TO OPTICAL POWER
# ==================================================

power = a * voltage + b

# Power in mW
# Convert to W if required:
power_W = power / 1000


# ==================================================
# 5. CALCULATE STATISTICS
# ==================================================

maximum_power = np.max(power)
minimum_power = np.min(power)
average_power = np.mean(power)
std_power = np.std(power)

rms_power = np.sqrt(np.mean(power**2))

print("\n========== OPTICAL POWER ANALYSIS ==========")

print(f"Maximum power: {maximum_power:.5f} mW")
print(f"Minimum power: {minimum_power:.5f} mW")
print(f"Average power: {average_power:.5f} mW")
print(f"Standard deviation: {std_power:.5f} mW")
print(f"RMS power: {rms_power:.5f} mW")

if average_power != 0:
    print(f"Relative standard deviation: {100 * std_power / abs(average_power):.3f} %")


# ==================================================
# 6. PUBLICATION-QUALITY PLOT
# ==================================================

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 14,
    "axes.labelsize": 18,
    "axes.titlesize": 20,
    "xtick.labelsize": 14,
    "ytick.labelsize": 14,
    "axes.linewidth": 1.5,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
})

fig, ax = plt.subplots(figsize=(12, 7))

ax.plot(time, power, color="red", linewidth=1.0, label="Optical power")

ax.set_xlabel("Time (s)")
ax.set_ylabel("Optical Power (W)")

ax.set_title("Short term optical Power stability")

ax.grid(True, linestyle="--", linewidth=0.6, alpha=0.4)

ax.legend(fontsize=13, frameon=True, loc="best")

plt.tight_layout()

# Save 600 DPI PNG
plt.savefig("Optical_Power_vs_Time_600dpi.png", dpi=600, bbox_inches="tight")

# Save vector PDF
plt.savefig("Optical_Power_vs_Time.pdf", bbox_inches="tight")

# Save vector SVG
plt.savefig("Optical_Power_vs_Time.svg", bbox_inches="tight")

plt.show()


# ==================================================
# 7. SAVE CONVERTED DATA
# ==================================================


print("\nAnalysis completed successfully!")

print("Files generated:")
print("1. Optical_Power_vs_Time_600dpi.png")
print("2. Optical_Power_vs_Time.pdf")
print("3. Optical_Power_vs_Time.svg")
print("4. Converted_Optical_Power_Data.csv")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
from matplotlib.patches import Rectangle

# ==================================================
# 1. CALIBRATION PARAMETERS
# ==================================================

a = 2.3361
b = -0.1964

# Calibration:
# P (mW) = a * V_PD + b

# ==================================================
# 2. ZOOM REGION SETTINGS
# ==================================================

zoom_start = 50.0  # Starting time (s)
zoom_end = 75.0  # Ending time (s)

# ==================================================
# 3. LOAD OSCILLOSCOPE CSV FILE
# ==================================================

filename = "/home/ae19663/Desktop/Deepak/02-11-2026/aom switch on for lon after switch off_000.csv"

with open(filename, "r", encoding="utf-8-sig") as f:
    lines = f.readlines()

# Locate TIME and CH3 header
header_row = None

for i, line in enumerate(lines):
    if "TIME" in line and "CH3" in line:
        header_row = i
        break

if header_row is None:
    raise ValueError("TIME and CH3 columns not found in CSV.")

print("Header found at row:", header_row)

# ==================================================
# 4. READ CSV DATA
# ==================================================

df = pd.read_csv(filename, skiprows=header_row, sep=None, engine="python")

df.columns = df.columns.str.strip()

print("\nCSV columns:")
print(df.columns.tolist())

# Extract time and voltage
time = pd.to_numeric(df["TIME"], errors="coerce")
voltage = pd.to_numeric(df["CH3"], errors="coerce")

# Remove invalid values
valid = time.notna() & voltage.notna()

time = time[valid].to_numpy()
voltage = voltage[valid].to_numpy()

# Sort by time
sort_index = np.argsort(time)

time = time[sort_index]
voltage = voltage[sort_index]

print("\nNumber of samples:", len(time))

# ==================================================
# 5. CONVERT VOLTAGE TO OPTICAL POWER
# ==================================================

power_mW = a * voltage + b

# Convert mW to W
power_W = power_mW 

# ==================================================
# 6. CALCULATE STATISTICS
# ==================================================

maximum_power = np.max(power_W)
minimum_power = np.min(power_W)
average_power = np.mean(power_W)
std_power = np.std(power_W)

rms_power = np.sqrt(np.mean(power_W**2))

print("\n========== OPTICAL POWER ANALYSIS ==========")

print(f"Maximum power: {maximum_power:.6f} W")
print(f"Minimum power: {minimum_power:.6f} W")
print(f"Average power: {average_power:.6f} W")
print(f"Standard deviation: {std_power:.6f} W")
print(f"RMS power: {rms_power:.6f} W")

if average_power != 0:
    print(f"Relative standard deviation: {100 * std_power / abs(average_power):.3f} %")

# ==================================================
# 7. PUBLICATION-QUALITY PLOT SETTINGS
# ==================================================

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 14,
    "axes.labelsize": 18,
    "axes.titlesize": 20,
    "xtick.labelsize": 13,
    "ytick.labelsize": 13,
    "axes.linewidth": 1.5,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
})

# ==================================================
# 8. MAIN OPTICAL POWER PLOT
# ==================================================

fig, ax = plt.subplots(figsize=(12, 7))

ax.plot(time, power_W, color="red", linewidth=1.0, label="Optical power")

ax.set_xlabel("Time (s)")
ax.set_ylabel("Optical Power (W)")

ax.set_title("Optical Power of Dimple Trap vs Time")

ax.grid(True, linestyle="--", linewidth=0.6, alpha=0.4)

ax.legend(fontsize=13, frameon=True, loc="best")

# ==================================================
# 9. ZOOMED INSET AT BOTTOM RIGHT
# ==================================================

# Select zoomed time interval
mask = (time >= zoom_start) & (time <= zoom_end)

if not np.any(mask):
    raise ValueError(
        "No data found in the selected zoom interval. "
        "Please change zoom_start and zoom_end."
    )

zoom_time = time[mask]
zoom_power = power_W[mask]

# Create inset
axins = inset_axes(ax, width="38%", height="38%", loc="lower right", borderpad=2)

# Plot zoomed data
axins.plot(zoom_time, zoom_power, color="red", linewidth=1.0)

# Determine inset Y limits
y_min = np.min(zoom_power)
y_max = np.max(zoom_power)

y_padding = max((y_max - y_min) * 0.15, 1e-9)

y_low = y_min - y_padding
y_high = y_max + y_padding

axins.set_xlim(zoom_start, zoom_end)
axins.set_ylim(y_low, y_high)

# Inset title
axins.set_title("Zoomed region", fontsize=11, pad=5)

# Inset grid
axins.grid(True, linestyle="--", linewidth=0.5, alpha=0.4)

axins.tick_params(axis="both", labelsize=10, direction="in")

# ==================================================
# 10. HIGHLIGHT ZOOM REGION ON MAIN PLOT
# ==================================================

zoom_rectangle = Rectangle(
    (zoom_start, y_low),
    zoom_end - zoom_start,
    y_high - y_low,
    linewidth=1.2,
    edgecolor="black",
    facecolor="none",
    zorder=5,
)

ax.add_patch(zoom_rectangle)

# No diagonal connector lines

# ==================================================
# 11. SAVE HIGH-RESOLUTION FIGURES
# ==================================================

plt.tight_layout()

# 600 DPI PNG
plt.savefig("Optical_Power_vs_Time_600dpi.png", dpi=600, bbox_inches="tight")

# Vector PDF
plt.savefig("Optical_Power_vs_Time.pdf", bbox_inches="tight")

# Vector SVG
plt.savefig("Optical_Power_vs_Time.svg", bbox_inches="tight")

plt.show()

# ==================================================
# 12. SAVE CONVERTED DATA
# ==================================================

output = pd.DataFrame({
    "Time (s)": time,
    "Voltage (V)": voltage,
    "Optical Power (mW)": power_mW,
    "Optical Power (W)": power_W,
})

output.to_csv("Converted_Optical_Power_Data.csv", index=False)

print("\nAnalysis completed successfully!")

print("\nFiles generated:")
print("1. Optical_Power_vs_Time_600dpi.png")
print("2. Optical_Power_vs_Time.pdf")
print("3. Optical_Power_vs_Time.svg")
print("4. Converted_Optical_Power_Data.csv")
```
