This is designed to extract params from previious experiment runs for future reference.RuntimeError

We have to variously parse them and its a pita to do, but we only pull the params that arent their defaults to avoid massive lists.

```python
import h5py
from sipyco import pyon

def get_override_args(filename, hide_defaults=True):
    vals = {}

    f1 = h5py.File(filename, "r")
    args = pyon.decode(f1["expid"][()])["arguments"]
    for k, v in args.items():
        if k == "ndscan_params":
            params = pyon.decode(args["ndscan_params"])
            overrides = params["overrides"]
            for k, v in overrides.items():
                spec = params["schemata"][k]["spec"]
                val = v[0]["value"]
                default = type(val)(params["schemata"][k]["default"])
                if hide_defaults and val == default:
                    continue
                vals[k] = [val / spec.get("scale", 1), spec.get("unit", "")]
        else:
            vals[k] = [v, ""]
    return vals
```

```python
vals = get_override_args("../results/2025-07-05/18/000006447-AbsorptionImageExpFrag.h5")

for k, (v, unit) in vals.items():
    print(f"{k} = {v} {unit}")
```
