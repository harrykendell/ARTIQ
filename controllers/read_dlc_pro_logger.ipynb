```python
import matplotlib.pyplot as plt
import h5py

directory="/home/ae19663/Desktop/toptica_dlc_pro/2026-06-13_15-08-53"

with h5py.File(f"{directory}/experiment.h5", "r") as f:
    shot = f["shot_000001"]
    #print number of shots in the file
    print("Number of shots in the file: ", len(f.keys()))

    #print timestamp of the all shots in the file
    for key in f.keys():
        print("Timestamp of shot: ", key, " is: ", f[key].attrs["timestamp"])

    x = shot["scope"]["x"][:]
    y = shot["scope"]["y"][:]

plt.plot(x, y)
plt.title("Scope Data")
plt.xlabel("piezo voltage [V]")
plt.ylabel("photodiode voltage [V]")
plt.grid()
plt.show()
```
